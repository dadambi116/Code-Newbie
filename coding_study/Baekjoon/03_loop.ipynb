# 25304 영수증

```python
X = int(input())
N = int(input())
list = []
for _ in range(N):
    A, B = map(int, input().split(' '))
    list.append(A*B)
    for i in range(len(list)):
        result = 0
        result += list[i]
        print(result)

# 값이 하나씩 나옴  
# 범위를 잘 생각해야!
# result의 초기값이 계속 반복되는 문제 발생
# 거창하게 중복 for문을 사용할 필요 없음!
    
```

```python
# 수정
X = int(input())
N = int(input())
result = 0
for _ in range(N):
    A, B = map(int, input().split())
    result += A * B
if result == X:
    print("Yes")
else:
    print("No")

```

# 25314 코딩은 체육과목입니다

```python
N = int(input())
for i in range(N//4):
    result = 'long'
    print(result, end=' ')
print('int')
```

```python
n = int(input())
result = ""  # 결과를 저장할 빈 문자열

# n을 4로 나눈 만큼 반복
for i in range(n//4):
    result += "long "  # long과 공백 추가

result += "int"  # 마지막에 int 추가
print(result)
```

15552 빠른 A + B

```python
import sys
T = int(sys.stdin.readline())
for i in range(T):
    a, b = map(int, sys.stdin.readline().split())
    print(a + b)
```

```python
import sys

T = int(sys.stdin.readline())
# sys.stdout.write를 사용하면 더 빠른 출력이 가능합니다
# 문자열 출력만 가능!
for _ in range(T):
    A, B = map(int, sys.stdin.readline().split())
    sys.stdout.write(str(A + B) + '\n')
```
