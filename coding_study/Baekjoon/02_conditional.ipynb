```python
# 같은 수가 몇개 있는지 확인하는 코드!
from collections import Counter

numbers = [2,3,4,2,6,3]

count = Counter(numbers)

for num, cnt in count.items():
    if cnt > 1:
        print(f"숫자 {num}는 {cnt}번 등장") 


# import 없이 
numbers = [2,3,4,2,6,3]
count = {}  # 딕셔너리 설정 중요!
            # count[key] = value 이렇게 호환됨

# 숫자 횟수 count
for num in numbers:
    if num in count:
        count[num] += 1
    else:
        count[num] = 1
print(count)
```

```python
# GPT 코드 
# 세 개의 숫자(주사위 눈) 입력 받기
A, B, C = map(int, input().split())

# 숫자들을 리스트로 저장
numbers = [A, B, C]

# 중복된 숫자의 개수를 세기 위해 set을 사용
unique_numbers = list(set(numbers))

if len(unique_numbers) == 1:
    # 세 눈이 모두 같을 경우
    result = 10000 + A * 1000
    print(result)
elif len(unique_numbers) == 2:
    # 두 눈이 같을 경우
    if numbers.count(A) == 2:
        result = 1000 + A * 100
    else:
        result = 1000 + B * 100
    print(result)
else:
    # 모두 다른 눈일 경우
    result = max(numbers) * 100
    print(result)
```
