- 1085 직사각형에서 탈출

```python
x, y, w, h = map(int,input().split())


if abs(x - w) < x:
    result1 = abs(x - w)
else:
    result1 = x

if abs(y - h) < y:
    result2 = abs(y - h)
else:
    result2 = y

print(min(result1, result2))
```

- 1247 부호

```python
for _ in range(3):
    num = int(input())
    sum = 0
    for i in range(num):
        a = int(input())
        sum += a
    if sum == 0:
        print(0)
    elif sum > 0:
        print('+')
    else:            
        print("-")
```

- 1284 집 주소

```python
address = []

while True:
    address = input()

    if address == '0':
        break
    
    sum = 0
    for j in range(len(address)):
        if address[j] == '0':
            sum += 4
        elif address[j] == '1':
            sum += 2
        else:
            sum += 3
    sum += len(address) + 1
    print(sum)
```

- 1547 공

```python
num = int(input())
cup = [0, 1, 2, 3]
for i in range(num):
    a, b = map(int, input().split())
    cup[a], cup[b] = cup[b], cup[a]
print(cup.index(1))
```
