- 1264 모음의 개수

```python
vowels = ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']
# count = 0 -> 오답인 이유?
while True:
    count = 0
    sentences = input()
    if sentences == '#':
        break
    for sentence in sentences:
        if sentence in vowels:
            count += 1
    print(count)
```

- 2530 인공지능 시계

```python
A, B, C = map(int, input().split())
D = int(input())

C = C + D % 60
B = B + D // 60

if C >= 60:
    C = C - 60
    B += 1
    if B >= 60:
        B = B - 60
        A += 1
    else:
        B = B
else: 
    C = C
print(f'{A} {B} {C}')
```

```python
# 10101 삼각형 외우기
a = int(input())
b = int(input())
c = int(input())

if a == b == c == 60:
    print('Equilateral')
elif a == b or b == c or a == c and sum(a, b, c) == 180:
    print('Isosceles')
elif sum(a, b, c) == 180 and a!= b and b!= c and a!= c:
    print("Scalence")
else:
    print("Error")
```
