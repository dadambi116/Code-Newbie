## 10810 공 넣기

```python
N, M = map(int,input().split())
n_list = [0] * N  # 아무것도 안 들어갈 경우 초기값  / n개의 바구니 생성
for _ in range(M):
    i, j, k = list(map(int,input().split()))
    for a in range(i, j+1):
        n_list[a-1] = k # 리스트의 첫번재(인덱스 0) 자리에 K 들어감

# 출력
for x in range(N): 
    print(n_list[x], end=' ') 
```

- 10813 공 바꾸기

```python
N, M = map(int, input().split()) 
basket = [] 
for x in range(N):
    basket.append(x+1)   # 바구니 생성 , 공 넣기 (인덱스 값 +1)
# print(basket)
for _ in range(M): # 공 바꾸기
    i, j = map(int, input().split())
    k =  basket[i-1]
    z = basket[j-1]
    basket[j-1] = k 
    basket[i-1] = z 
print(*basket)
        
```

```python
N, M = map(int, input().split()) 
basket = [] 
for x in range(N):
    basket.append(x+1)   # 바구니 생성 , 공 넣기 (인덱스 값 +1)
# print(basket)
for _ in range(M): # 공 바꾸기
    i, j = map(int, input().split())
    basket[i-1], basket[j-1] = basket[j-1], basket[i-1] 
print(*basket)
```

- 10811 바구니 뒤집기

```python
N, M = map(int, input().split())
basket = [_ for _ in range(1, N+1)] 
for _ in range(M):
    i, j = map(int, input().split())
    reverse_b = basket[i-1:j] # 바꿀 바구니의 범위 지정
    reverse_b.reverse()         # 역순으로 바꿈
    basket[i-1:j] = reverse_b   # 바꾼 배열을 원본에 저장!

for k in range(N):  
    print(basket[k], end=' ') # 인덱스 하나씩 출력
```

- 1546 평균

```python
N = int(input())
score = list(map(int, input().split()))
M = max(score)

for i in range(N):
    score[i] = score[i]/M*100
print(sum(score)/N)
```

```python
# 다른 풀이 찾아본 거
N = int(input())
score = list(map(int, input().split()))
M = max(score)
sum = 0

for i in range(N):
    score[i] = score[i]/M*100
    sum += score[i]

avg = sum/N
print(avg)
```
