- 10809번 알파벳 찾기

```python
s = input()
alphbet = 'abcdefghijklmnopqrstuvwxyz'

for i in alphbet:
    if i in s:
        print(s.index(i), end=' ')
    else:
        print('-1', end=' ')
```
